### imports

```python
import sys
import os
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy as sci
from scipy.stats import qmc
import xarray as xr

wrkdir = "/scratch/es60/pjb581/access-om3/archive/"

# print versions of packages
print("python version =",sys.version[:5])
print("numpy version =", np.__version__)
print("pandas version =", pd.__version__)
print("scipy version =", sci.__version__)
print("matplotlib version =", sys.modules[plt.__package__].__version__)

os.chdir(wrkdir)
#os.listdir()
```

### load experiments

```python
%%time

vari = ['dic_stf', 'bac1', 'bac2', 'bac2_rq']

exps = np.arange(1,401,1)
ds = {}
for exp in exps:
    if exp % 20 == 0:
        print(exp)
    ds["%s_%i"%(vari[0],exp)] = xr.open_dataset('om3_mom6_1D-SOTS_exp%i/output009/access-om3.mom6.2d.%s.1day.mean.1909.nc'%(exp,vari[0]), decode_timedelta=True)[vari[0]]
    ds["%s_%i"%(vari[1],exp)] = xr.open_dataset('om3_mom6_1D-SOTS_exp%i/output009/access-om3.mom6.3d.%s.1day.mean.1909.nc'%(exp,vari[1]), decode_timedelta=True)[vari[1]]
    ds["%s_%i"%(vari[2],exp)] = xr.open_dataset('om3_mom6_1D-SOTS_exp%i/output009/access-om3.mom6.3d.%s.1day.mean.1909.nc'%(exp,vari[2]), decode_timedelta=True)[vari[2]]
    ds["%s_%i"%(vari[3],exp)] = xr.open_dataset('om3_mom6_1D-SOTS_exp%i/output009/access-om3.mom6.3d.%s.1day.mean.1909.nc'%(exp,vari[3]), decode_timedelta=True)[vari[3]]
```

### partial oxidation experiment inputs

```python
ds["%s_%i"%(vari[1],exp)]
```

```python
%%time

bac1_alpha = np.arange(0.0, 0.951, 0.05)
bac2_alpha = np.arange(0.0, 0.951, 0.05)

bac1_a, bac2_a = np.meshgrid(bac1_alpha, bac2_alpha)

plt.figure()
plt.pcolormesh(bac1_alpha, bac2_alpha, bac1_a*bac2_a)
```

### calculate means over annual cycle

```python
%%time

co2_flux = np.zeros((400))
bac1 = np.zeros((400))
bac2 = np.zeros((400))
bac2_rq = np.zeros((400))

for exp in exps:
    co2_flux[exp-1] = ds["%s_%i"%(vari[0],exp)].isel(yh=0,xh=0).mean(dim='time') * 365 * 86400
    bac1[exp-1] = ds["%s_%i"%(vari[1],exp)].isel(yh=0,xh=0,zl=0).mean(dim='time') * 1e6 * 1.035
    bac2[exp-1] = ds["%s_%i"%(vari[2],exp)].isel(yh=0,xh=0,zl=0).mean(dim='time') * 1e6 * 1.035
    bac2_rq[exp-1] = ds["%s_%i"%(vari[3],exp)].isel(yh=0,xh=0,zl=0).mean(dim='time')

co2_flux = co2_flux.reshape(20,20)
bac1 = bac1.reshape(20,20)
bac2 = bac2.reshape(20,20)
bac2_rq = bac2_rq.reshape(20,20)
```

### plot CO2 fluxes

```python

plt.figure()
plt.contourf(bac1_alpha, bac2_alpha, co2_flux, levels=np.arange(-0.2,4.1,0.2), extend='both')
c1 = plt.colorbar()
plt.contour(bac1_alpha, bac2_alpha, co2_flux, levels=[0.0], colors='goldenrod')
plt.contour(bac1_alpha, bac2_alpha, co2_flux, levels=[1.0, 2.0, 3.0, 4.0], colors='firebrick')
plt.xlabel("Particle associated bacteria\ndegree of partial oxidation")
plt.ylabel("Free-living bacteria\ndegree of partial oxidation")
c1.ax.set_ylabel("Annual mean CO$_2$ flux")

plt.figure()
plt.contourf(bac1_alpha, bac2_alpha, bac1, levels=np.arange(0.0,0.51,0.025), extend='max')
c1 = plt.colorbar()
#plt.contour(bac1_alpha, bac2_alpha, bac1, levels=[0.0], colors='goldenrod')
#plt.contour(bac1_alpha, bac2_alpha, bac1, levels=[1.0, 2.0, 3.0, 4.0], colors='firebrick')
plt.xlabel("Particle associated bacteria\ndegree of partial oxidation")
plt.ylabel("Free-living bacteria\ndegree of partial oxidation")
c1.ax.set_ylabel("Annual mean surface particle-associated bacteria")

plt.figure()
plt.contourf(bac1_alpha, bac2_alpha, bac2, levels=np.arange(0.0,0.51,0.025), extend='max')
c1 = plt.colorbar()
#plt.contour(bac1_alpha, bac2_alpha, bac2, levels=[0.0], colors='goldenrod')
#plt.contour(bac1_alpha, bac2_alpha, bac2, levels=[1.0, 2.0, 3.0, 4.0], colors='firebrick')
plt.xlabel("Particle associated bacteria\ndegree of partial oxidation")
plt.ylabel("Free-living bacteria\ndegree of partial oxidation")
c1.ax.set_ylabel("Annual mean surface free-living bacteria")

plt.figure()
plt.contourf(bac1_alpha, bac2_alpha, bac2_rq, levels=np.arange(0.3,1.21,0.05), extend='both')
c1 = plt.colorbar()
#plt.contour(bac1_alpha, bac2_alpha, bac2_rq, levels=[0.0], colors='goldenrod')
#plt.contour(bac1_alpha, bac2_alpha, bac2_rq, levels=[1.0, 2.0, 3.0, 4.0], colors='firebrick')
plt.xlabel("Particle associated bacteria\ndegree of partial oxidation")
plt.ylabel("Free-living bacteria\ndegree of partial oxidation")
c1.ax.set_ylabel("Annual mean surface RQ")
```
