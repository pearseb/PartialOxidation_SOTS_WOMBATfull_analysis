```python
%%time

# Use the same parameter set
X.columns = list(range(35))
X_O = X

tarcols = cost.columns[0:-1]
print(tarcols)

# Use the cost function defined earlier == (1-r)*NRMSE
Y_O = np.zeros((2048, len(tarcols)))
for i, target in enumerate(tarcols):
    Y_O[:,i] = cost[target]; suffix = 'cost'
    #y_O[:,i] = rmse[target]; suffix = 'rmse'
    #Y_O[:,i] = corr[target]; suffix = 'corr'
    # Normalising Error values (Error - min(Error)/max(Error) - min(Error))
    Y_O[:,i]=(Y_O[:,i]-Y_O[:,i].min())/(Y_O[:,i].max()-Y_O[:,i].min())
```

### Train the GPR on the cost function from the 2048 model samples

```python
n_vars   = len(tarcols)
n_params = len(X.columns)
N        = len(X)

print(n_vars, n_params, N)
```

```python
os.getcwd()
```

```python
%%time

import warnings
warnings.filterwarnings('ignore')

fold = int(8)
fold_size = int(N/fold)

# --- define the GPR model ---
def make_gpr():
    kernel = (
        ConstantKernel(1.0, (1e-3, 1e3))
        * Matern(
            length_scale=np.ones(n_params),
            length_scale_bounds=(1e-3, 5.0),
            nu=1.5
        )
        + WhiteKernel(
            noise_level=1e-6,
            noise_level_bounds=(1e-9, 1e-2)
        )
    )
    model = GaussianProcessRegressor(
        kernel=kernel,
        n_restarts_optimizer=5,
        random_state=42
    )
    return model

# --- worker function ---
def fit_fold(i, k):
    import warnings
    warnings.filterwarnings("ignore")

    model = make_gpr()

    X_train = pd.concat(
        [X_O.iloc[0:k*fold_size, :], X_O.iloc[(k+1)*fold_size:, :]],
        axis=0
    )

    X_test = X_O.iloc[k*fold_size:(k+1)*fold_size, :]

    y_train = np.concatenate(
        [Y_O[0:k*fold_size, i], Y_O[(k+1)*fold_size:, i]],
        axis=0
    )

    y_test = Y_O[k*fold_size:(k+1)*fold_size, i]

    model.fit(X_train, y_train)
    y_hat = model.predict(X_test)

    r2 = r2_score(y_test, y_hat)

    return i, k, y_hat, y_test, r2

```

```python
%%time

# --- parallel execution across all targets and folds ---
with parallel_backend("dask"):
    results_optim = Parallel(prefer="processes", verbose=10)(
        delayed(fit_fold)(i, k)
        for i in range(n_vars)
        for k in range(fold)
    )
```

```python
%%time

GPR_score = np.zeros((fold, n_vars))
Y_PRED = np.zeros((N, n_vars))
Y_TRUE = np.zeros((N, n_vars))

for i, k, y_hat, y_test, r2 in results_optim:
    idx0 = k * fold_size
    idx1 = (k + 1) * fold_size

    Y_PRED[idx0:idx1, i] = y_hat
    Y_TRUE[idx0:idx1, i] = y_test
    GPR_score[k, i] = r2

r2_summary = pd.DataFrame({
    "target": tarcols,
    "mean_fold_R2": np.mean(GPR_score, axis=0),
    "std_fold_R2": np.std(GPR_score, axis=0),
    "overall_CV_R2": [
        r2_score(Y_TRUE[:, i], Y_PRED[:, i])
        for i in range(n_vars)
    ]
})

print(r2_summary)

```

### save the models

```python
%%time

def fit_final_model(i):
    target = tarcols[i]
    model = make_gpr()

    y = Y_O[:, i]

    model.fit(X_O, y)

    y_hat = model.predict(X_O)
    train_r2 = r2_score(y, y_hat)

    filename = f"GPR_{target}_{suffix}.joblib"
    joblib.dump(model, filename)

    return {
        "target": target,
        "metric_emulated": suffix,
        "filename": filename,
        "training_R2": train_r2,
        "kernel": str(model.kernel_)
    }

with parallel_backend("dask"):
    final_results = Parallel(prefer="processes", verbose=10)(
        delayed(fit_final_model)(i)
        for i in range(n_vars)
    )

final_summary = pd.DataFrame(final_results)
print(final_summary)
```

```python
%%time

#----------------------------------Bayesian Optimisation----------------------------------
#                        Written by P. Jyoteeshkumar Reddy (2024)
#

def lnprior(theta):
    if not np.all((theta >= 0) & (theta <= 1)):
        return -np.inf
    prior_sigma = 0.35
    return -0.5 * np.sum(((theta - 0.5) / prior_sigma) ** 2)


def param_fill(sens_params, theta, fixed_params=None, fixed_values=None):
    all_params = ['P1', 'P2', 'P3','P4', 'P5', 'P6','P7', 'P8', 'P9','P10',
                  'P11', 'P12', 'P13', 'P14', 'P15', 'P16', 'P17', 'P18', 'P19', 'P20',
                  'P21', 'P22', 'P23', 'P24', 'P25', 'P26', 'P27', 'P28', 'P29', 'P30',
                  'P31', 'P32', 'P33', 'P34', 'P35']
    sens_indices = [all_params.index(param) for param in sens_params]
    sample = np.full(len(all_params), 0.5)
    sample[sens_indices] = theta
    if fixed_params is not None and fixed_values is not None:
        fixed_indices = [all_params.index(param) for param in fixed_params]
        sample[fixed_indices] = fixed_values
    return sample


def lnprob(theta):
    lp = lnprior(theta)
    if not np.isfinite(lp):
        return -np.inf
    return lp + lnlike(theta)


def optimise_checkpointed(
    ndim,
    nsteps_total,
    theta_best=None,
    nwalkers=None,
    nproc=56,
    checkpoint_file="mcmc_checkpoint.h5",
    chunk_size=50000,
    reset=False
):
    if nwalkers is None:
        nwalkers = max(4 * ndim, 100)

    # Create backend
    backend = emcee.backends.HDFBackend(checkpoint_file)

    # Either start fresh or resume
    if reset:
        backend.reset(nwalkers, ndim)

        if theta_best is None:
            pos = np.random.rand(nwalkers, ndim)
        else:
            theta_best = np.asarray(theta_best)

            if len(theta_best) != ndim:
                raise ValueError(f"theta_best has length {len(theta_best)}, but ndim = {ndim}")

            pos = theta_best + 1e-3 * np.random.randn(nwalkers, ndim)
            pos = np.clip(pos, 1e-6, 1 - 1e-6)

        already_done = 0

    else:
        already_done = backend.iteration

        if already_done == 0:
            if theta_best is None:
                pos = np.random.rand(nwalkers, ndim)
            else:
                theta_best = np.asarray(theta_best)

                if len(theta_best) != ndim:
                    raise ValueError(f"theta_best has length {len(theta_best)}, but ndim = {ndim}")

                pos = theta_best + 1e-3 * np.random.randn(nwalkers, ndim)
                pos = np.clip(pos, 1e-6, 1 - 1e-6)
        else:
            pos = None  # resume from backend

    print(f"Checkpoint file: {checkpoint_file}")
    print(f"Already completed: {already_done} steps")
    print(f"Target total: {nsteps_total} steps")

    remaining = nsteps_total - already_done

    if remaining <= 0:
        print("Already reached requested number of steps.")
        chains = backend.get_chain()
        log_probs = backend.get_log_prob()

        try:
            tau = backend.get_autocorr_time()
            print("Autocorrelation time:", tau)
        except emcee.autocorr.AutocorrError as e:
            tau = e.tau
            print("WARNING: Chain is shorter than 50 times the estimated autocorrelation time.")
            print("Estimated autocorrelation time:", tau)

        return backend, chains, log_probs, tau

    with Pool(processes=nproc) as pool:
        moves = [
            (emcee.moves.StretchMove(a=1.3), 0.7),
            (emcee.moves.DEMove(), 0.3),
        ]
        
        sampler = emcee.EnsembleSampler(
            nwalkers,
            ndim,
            lnprob,
            pool=pool,
            backend=backend,
            moves=moves
        )

        n_left = remaining

        while n_left > 0:
            this_chunk = min(chunk_size, n_left)

            print(f"Running next chunk: {this_chunk} steps")

            sampler.run_mcmc(
                pos,
                this_chunk,
                progress=False
            )

            # After first chunk, sampler continues internally
            pos = None

            n_left -= this_chunk

            completed = backend.iteration
            print(f"Completed: {completed}/{nsteps_total} steps")

            # Optional intermediate diagnostic
            try:
                tau_now = backend.get_autocorr_time()
                acceptance = sampler.acceptance_fraction
                print("Mean acceptance fraction:", np.mean(acceptance))
                print("Min acceptance fraction:", np.min(acceptance))
                print("Max acceptance fraction:", np.max(acceptance))
                print("Current autocorrelation time:", tau_now)
                print("Current max tau:", np.max(tau_now))
                print("Steps / max(tau):", completed / np.max(tau_now))
            except emcee.autocorr.AutocorrError as e:
                tau_now = e.tau
                acceptance = sampler.acceptance_fraction
                print("Mean acceptance fraction:", np.mean(acceptance))
                print("Min acceptance fraction:", np.min(acceptance))
                print("Max acceptance fraction:", np.max(acceptance))
                print("Current estimated tau:", tau_now)
                print("Current max tau:", np.max(tau_now))
                print("Steps / max(tau):", completed / np.max(tau_now))

    chains = backend.get_chain()
    log_probs = backend.get_log_prob()

    try:
        tau = backend.get_autocorr_time()
        print("Final autocorrelation time:", tau)
    except emcee.autocorr.AutocorrError as e:
        tau = e.tau
        print("WARNING: Chain is shorter than 50 times the estimated autocorrelation time.")
        print("Final estimated autocorrelation time:", tau)

    # HDFBackend does not directly store acceptance fraction as persistently
    # as the in-memory sampler object, so return sampler too if useful.
    acceptance = np.mean(sampler.acceptance_fraction)
    print("Mean acceptance fraction:", acceptance)

    return sampler, chains, log_probs, tau


def pmp(data):
    data = np.asarray(data)
    if np.std(data) == 0:
        return data[0]
    kde = gaussian_kde(data)
    x_grid = np.linspace(data.min(), data.max(), 1000)
    kde_values = kde(x_grid)
    mode_value = x_grid[np.argmax(kde_values)]
    return mode_value

```

### Based on above sensitivity analysis, select which parameters are important

```python
%%time

### Only optimise certain sensitive parameters
sens_params = ['P1',  'P2',  'P3',  'P5', 'P6', 'P7', 'P8', 'P10',
               'P11', 'P12', 'P13', 'P16', 'P18',
               'P21', 'P22', 'P23', 'P24', 'P25', 'P26', 'P27', 'P28', 'P30', 
               'P31', 'P32', 'P33', 'P34', 'P35']
safe_params = ['P4', 'P9', 'P14', 'P15', 'P17', 'P19', 'P20', 'P29']
safe_values = np.ones(len(safe_params))*0.5  # whatever fixed values you want

def lnlike(theta):
    full_theta = param_fill(sens_params, theta, safe_params, safe_values)
    models = [fgco2, no3, sil, dfe, poc, bac, micro, meso, chl_bottle, chl_mooring, b1mb2, D2L300mD2L5]
    loglike = 0.0
    for model in models:
        mu, std = model.predict([full_theta], return_std=True)
        err = max(mu[0], 0.0)
        sigma = np.sqrt(0.10**2 + std[0]**2)
        loglike += -0.5 * (err / sigma) ** 2 - np.log(sigma)
    return loglike

```

#### Do a quick search for a good starting point of initial values

```python
%%time

def neg_lnprob(theta):
    val = lnprob(theta)
    if not np.isfinite(val):
        return 1e30
    return -val

bounds = [(0, 1)] * len(sens_params)

result = differential_evolution(
    neg_lnprob,
    bounds,
    maxiter=1000,
    popsize=20,
    polish=True,
    seed=42
)

theta_best = result.x
theta_start = np.clip(theta_best, 0.05, 0.95)
```

```python
%%time

print("Best negative log-posterior:", result.fun)
print("Best log-posterior:", -result.fun)
print("Best theta:")
for i,p in enumerate(sens_params):
    print(p,theta_start[i], theta_best[i])
```

### draw 100 random samples from the posteriors

```python
%%time

n_samples = 100
rng = np.random.default_rng(42)

samples = pd.DataFrame(index=np.arange(n_samples), columns=all_params, dtype=float)
norm_samples = pd.DataFrame(index=np.arange(n_samples), columns=all_params, dtype=float)

for i, p in enumerate(all_params):

    low = ranges['min'].iloc[i] + (ranges['max'].iloc[i] - ranges['min'].iloc[i]) * ten_norm[i]
    mode = ranges['min'].iloc[i] + (ranges['max'].iloc[i] - ranges['min'].iloc[i]) * peak_norm[i]
    high = ranges['min'].iloc[i] + (ranges['max'].iloc[i] - ranges['min'].iloc[i]) * ninety_norm[i]

    # safety: ensure bounds are ordered
    low, high = min(low, high), max(low, high)

    # safety: triangular mode must sit inside [low, high]
    mode = np.clip(mode, low, high)
    mode_norm = np.clip(peak_norm[i], ten_norm[i], ninety_norm[i])

    samples[p] = rng.triangular(left=low, mode=mode, right=high, size=n_samples)
    norm_samples[p] = rng.triangular(left=ten_norm[i], mode=mode_norm, right=ninety_norm[i], size=n_samples)
    
```
