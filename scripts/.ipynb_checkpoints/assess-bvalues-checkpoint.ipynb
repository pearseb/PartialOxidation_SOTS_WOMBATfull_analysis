### Annika broken stick b-values

```python
%%time

# define spl model (simple power law)
def spl(x, b, cs):
    return (x / x[0]) ** (-b) * cs

# define bpl model (broken power law)
def bpl(x, b1, b2, cd, z_trans):
    return np.where(x < z_trans, (x/z_trans)**(-b1) * cd, (x/z_trans)**(-b2) * cd)

def r2_score_numpy(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)

    return 1 - ss_res / ss_tot
    
```

```python
depth_full = np.arange(0,1100,1)


exps = np.arange(1,401,1)
for exp in exps:
    if exp % 20 == 0:
        print(exp)
    ds["poc_interp_%i"%(exp)] = ds["poc_ave_%i"%(exp)].interp(zl=np.arange(0,1100,1))
    ds["mask_%i"%(exp)] = (depth_full > ds["zprod_ave_%i"%(exp)].values) & (depth_full < 1000.0)
    poc_ave = ds["poc_interp_%i"%(exp)][ds["mask_%i"%(exp)]]
    depth_ave = depth_full[ds["mask_%i"%(exp)]]

    # boundary and initial conditions for fit
    bounds_spl = ([-10,0], [10,150])
    bounds_bpl = ([-10, -10, 0, ds["zprod_ave_%i"%(exp)].values], [10, 10, 150, ds["zprod_ave_%i"%(exp)].values +300])
    p0_spl = ([0.3,50])
    p0_bpl = ([0.5, 0.3, 10, ds["zprod_ave_%i"%(exp)].values + 100])

    # fit spl
    popt_spl, _ = curve_fit(spl, depth_ave, poc_ave, nan_policy='omit', ftol=1e-3, sigma = 0.6*poc_ave, p0=p0_spl, bounds=bounds_spl)
    y_pred = spl(depth, *popt_spl)
    r2_spl = r2_score_numpy(poc, y_pred)

    # fit bpl
    # returns  b1, b2, cd, z_trans
    popt_bpl, _ = curve_fit(bpl, depth_ave, poc_ave, nan_policy='omit', ftol=1e-3, sigma = 0.6*poc_ave, p0=p0_bpl, bounds=bounds_bpl)
    y_pred = bpl(depth_ave, *popt_bpl)
    r2_bpl = r2_score_numpy(poc_ave, y_pred)

    # store results
    ds["b_spl_ave_%i"%(exp)] = popt_spl[0]
    ds["r2_spl_ave_%i"%(exp)] = r2_spl
    ds["b1_bpl_ave_%i"%(exp)] = popt_bpl[0]
    ds["b2_bpl_ave_%i"%(exp)] = popt_bpl[1]
    ds["ztrans_bpl_ave_%i"%(exp)] = popt_bpl[3]
    ds["r2_bpl_ave_%i"%(exp)] = r2_bpl

    ds["poc_interp_%i"%(exp)] = ds["poc_t365_%i"%(exp)].interp(zl=np.arange(0,1100,1))
    ds["mask_%i"%(exp)] = (depth_full > ds["zprod_t365_%i"%(exp)].values) & (depth_full < 1000.0)
    poc_t365 = ds["poc_interp_%i"%(exp)][ds["mask_%i"%(exp)]]
    depth_t365 = depth_full[ds["mask_%i"%(exp)]]

    # boundary and initial conditions for fit
    bounds_spl = ([-10,0], [10,150])
    bounds_bpl = ([-10, -10, 0, ds["zprod_t365_%i"%(exp)].values], [10, 10, 150, ds["zprod_t365_%i"%(exp)].values +300])
    p0_spl = ([0.3,50])
    p0_bpl = ([0.5, 0.3, 10, ds["zprod_t365_%i"%(exp)].values + 100])

    # fit spl
    popt_spl, _ = curve_fit(spl, depth_t365, poc_t365, nan_policy='omit', ftol=1e-3, sigma = 0.6*poc_t365, p0=p0_spl, bounds=bounds_spl)
    y_pred = spl(depth_t365, *popt_spl)
    r2_spl = r2_score_numpy(poc_t365, y_pred)

    # fit bpl
    # returns  b1, b2, cd, z_trans
    popt_bpl, _ = curve_fit(bpl, depth_t365, poc_t365, nan_policy='omit', ftol=1e-3, sigma = 0.6*poc_t365, p0=p0_bpl, bounds=bounds_bpl)
    y_pred = bpl(depth_t365, *popt_bpl)
    r2_bpl = r2_score_numpy(poc_t365, y_pred)

    # store results
    ds["b_spl_t365_%i"%(exp)] = popt_spl[0]
    ds["r2_spl_t365_%i"%(exp)] = r2_spl
    ds["b1_bpl_t365_%i"%(exp)] = popt_bpl[0]
    ds["b2_bpl_t365_%i"%(exp)] = popt_bpl[1]
    ds["ztrans_bpl_t365_%i"%(exp)] = popt_bpl[3]
    ds["r2_bpl_t365_%i"%(exp)] = r2_bpl
    
```
